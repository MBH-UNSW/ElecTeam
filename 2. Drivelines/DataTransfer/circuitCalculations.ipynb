```python
import math

## Primary Circuit LC Tank
A = 12.5    # voltage source amplitude
fc = 410e3  # carrier frequency
L = 15e-6   # inductor value

w = 2 * math.pi * fc    # angular frequency
C = 1 / (w**2 * L)      # required capacitor value

print("The required LC Tank Capacitor Value is (nF):", C / 1e-9)
print("The required LC Tank Capacitor Value is (pF):", C / 1e-12, "\n")


## Quality Factor
Rsc = 1     # equivalent series resistance of capacitor
Rsl = 0.53  # equivalent series resistance of inductor
R = 10      # resistance of primary circuit

Rs = Rsc + Rsl  # total equivalent series resistance
R_T = R + Rs    # total resistance of primary circuit

Q = math.sqrt(L/C) / R_T    # Q-Factor

print("The Q-Factor is:", Q, "\n")


## Secondary Circuit LC Tank
R1 = 1     # LC Tank resistor value
ZL = R1 + 1j * w * L    # impedance of R + L branch
ZC = 1 / (1j * w * C)   # impedance of parallel C branch

Ztank = (ZL * ZC) / (ZL + ZC)

# Rectifier Impedance
C2 = 1e-12  # rectifier capacitor value (IMPORTANT VALUE FOR CHANGE IN TOTAL IMPEDANCE)
ZC1 = 1 / (1j * w * C2) # impedance of C2
rd2 = 2.6   # resistance of diode

Zrectifier = rd2 + ZC1  # total impedance of recitifer

# Switch Impedance
rd1 = 2.6       # resistance of diode
RDS_ON = 2.6    # resistance of MOSFET

ZSCT = rd1 + RDS_ON     # total impedance of switch

# Total Impedance of Secondary Circuit
R2 = 1
alpha_1 = R2 + Zrectifier   # MOSFET OFF
alpha_2 = R2 + ZSCT         # MOSFET ON

Ztotal_OFF = (Ztank * alpha_1) / (Ztank + alpha_1)  # total impedance MOSFET OFF
Ztotal_ON = (Ztank * alpha_2) / (Ztank + alpha_2)   # total impedance MOSFET ON

Ztotal_OFF_abs = math.sqrt(Ztotal_OFF.real**2 + Ztotal_OFF.imag**2)
Ztotal_ON_abs = math.sqrt(Ztotal_ON.real**2 + Ztotal_ON.imag**2)

print("Total impedance OFF (ohms):", Ztotal_OFF_abs)
print("Total impedance ON (ohms):", Ztotal_ON_abs, "\n")


## Reflected Impedance
k = 0.05    # coupling coefficient
M = k * math.sqrt(L * L)    # mututal inductance (henries)
Zref_OFF = (w**2 * M**2) / (Ztotal_OFF)     # reflected impedance MOSFET OFF
Zref_ON = (w**2 * M**2) / (Ztotal_ON)       # reflected impedance MOSFET ON


## Primary Circuit with Reflected Impedance
XL = w * L          # inductor reactance
XC = 1 / (w * C)    # capacitor reactance
Zp = R_T + 1j* (XL - XC)    # total impedance of primary circuit

Ztotal_p_OFF = Zp + Zref_OFF
Ztotal_p_ON = Zp + Zref_ON

print("Total impedance primary OFF (ohms):", Ztotal_p_OFF)
print("Total impedance primary ON (ohms):", Ztotal_p_ON, "\n")

print("Total impedance primary OFF (ohms):", abs(Ztotal_OFF))
print("Total impedance primary ON (ohms):", abs(Ztotal_p_ON), "\n")


## R Range
Imax = 2    # max current allowed in primary circuit
R_required_OFF = (A / Imax) - Zref_OFF - Rs
R_required_ON = (A / Imax) - Zref_ON - Rs

R_required_OFF_abs = math.sqrt(R_required_OFF.real**2 + R_required_OFF.imag**2)
R_required_ON_abs = math.sqrt(R_required_ON.real**2 + R_required_ON.imag**2)

# print(R_required_OFF_abs)
# print(R_required_ON_abs)
```

```python
fc = 400e3  # carrier frequency

# Low Pass Filter
R = 1e3
lowPass_f = fc + 200e3
C = 1 / (2 * math.pi * lowPass_f * R)

print("The required capacitor value is (nF):", C / 1e-9)
print("The required capacitor value is (pF):", C / 1e-12, "\n")

# High Pass Filter
highPass_f = 2
C_hpf = 1 / (2* math.pi * highPass_f * R)

print("The required capacitor value is (mF):", C_hpf / 1e-3)
print("The required capacitor value is (uF):", C_hpf / 1e-6)
print("The required capacitor value is (nF):", C_hpf / 1e-9)
print("The required capacitor value is (pF):", C_hpf / 1e-12, "\n")

t_f = 1 / fc
t_hpf = R * C_hpf

print(t_f)
print(t_hpf)
print(t_hpf / t_f)
```
